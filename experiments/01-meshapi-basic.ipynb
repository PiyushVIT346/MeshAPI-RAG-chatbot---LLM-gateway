```python
import os 
from getpass import getpass
from dotenv import load_dotenv
load_dotenv()
```

```python
from meshapi import MeshAPI

MESHAPI_TOKEN = os.getenv("MESH_API_KEY")

client = MeshAPI(
    base_url = os.getenv("MESH_BASE_URL", "https://api.meshapi.ai"),
    token=MESHAPI_TOKEN
)

print("Client ready")
```

```python
client.models.list()
```

```python
len(client.models.list())
```

```python
ALL_MODELS_BY_NAME = {m.name: m for m in client.models.list()}
ALL_MODELS_BY_NAME[list(ALL_MODELS_BY_NAME.keys())[0]]
```

```python
from meshapi import ChatCompletionParams, ChatMessage

resp = client.chat.completions.create(
    ChatCompletionParams(
        model="google/gemini-2.5-flash",
        messages=[
            ChatMessage(
                role="user",
                content="In one sentence, what is an AI gateway ?"
            )
        ],
        max_tokens=300
    )
)

print(resp)
```

```python
usage = resp.usage
print(usage.prompt_tokens)
print(usage)
```
